```python
import yfinance as yf
#shortcut for use of yfinance
bmw = yf.download("BMW.DE", period = "5y")
#our first data pull of BMW stock data over the last 5 years
bmw.head()
#Shows us the first 5 rows of our data set
```

```python
import matplotlib.pyplot as plt
#package required for data visualisation of this kind
bmw["Close"].plot(figsize=(10,5), title = "BMW share price, last 5 years")
#bmw["Close"] selects the column we are interested in here, and the rest of this
#line outlines details of the graph itself
plt.ylabel("Price(EUR)")
#labelling the axis, notice the data labels the x-axis
plt.show()

```

```python
#OBSERVATIONS/TRENDS:
#Large increase across late 2022 to mid 2023 due to:
##large share buyback programme, electric vehicle growth and sales record

#Peak in Q1 2024 BECAUSE of:
##high EVs sales as well as ongoing market performance of the 5 Series

#Large decline across 2024 likely due to:
##large global vehicle recall due to faulty IBS (Integrated Braking System), severe sales slowdown

#General upward trend in 2025 likely due to:
##major executive transition, transition toward next-gen EVs - NEUE KLASSE, manufacturing innovations
```
