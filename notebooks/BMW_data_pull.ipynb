#Why profit and cash differ for BMW:
Profit and cash differ because profit is an accounting measure and cash is what actually moved through the bank account. For BMW, your table shows four causes.

1. Costs that don’t use cash

Depreciation (about €8.7bn a year at BMW) reduces profit but isn’t cash leaving that year. The cash left when the factory or the leased car was bought. So you add depreciation back, which pushes cash above profit.

2. Timing: working capital

Profit counts a sale when the car is delivered, but the cash may arrive later, or the money may be tied up in unsold stock. The table shows this clearly: change in working capital was -8.9bn in 2025 and -8.8bn in 2024, which cut cash by far more than the depreciation add-back could cover. In 2022 it was +3.0bn and helped cash.

BMW’s finance arm probably makes this bigger. When it lends to a car buyer, profit only picks up the interest over time, but the cash for the whole loan leaves on day one. I haven’t confirmed that this is what drives BMW’s figure, it'll be in the group cash flow statement (page 275 of the report) and the financial position section (from page 69), which show Automotive and Financial Services separately.

3. Tax and interest

The first row in your table is actually profit before tax, as we found. Cash tax has to come out of it, and so do other adjustments. That’s why simply adding depreciation and working capital to that first row doesn’t reproduce operating cash flow.
The remaining gap is “other items”, and in 2022 it’s large. That’s likely tax paid on a very profitable year plus other non-cash adjustments, but I’m inferring this.

4. Capital spending is not in profit at all

Buying factories, equipment and tooling (€11.0bn in 2025) never appears as a cost in profit, except gradually through depreciation. That’s why we subtract it to reach free cash flow, and it explains most of the swing from positive profit to negative free cash flow in the table.

#Revenue is down about 14% from the 2023 peak, so was profit hit mostly by lower sales or by lower margins?
Operating profit (EBIT) from 2023 to 2025 fell from €18.5bn to €10.2bn, down 45%, while revenue fell only 14%. To split the effect: if BMW had kept its 2023 margin of 11.9% on 2025's lower revenue, EBIT would have been about €15.9bn.
- Lower sales explains the fall from €18.5bn to €15.9bn, about €2.6bn, or roughly a third
- Lower margins explains the remaining drop from €15.9bn to €10.2bn, about €5.7bn or roughly two thirds.

So most of the damage came from earning less on each euro of sales, not from selling less. That fits what the report's early pages said: tariffs, tough competition in China and a heavier mix of costs. This is a good indicator of what needs to improve for BMW's profits to recover.

```python
balance = bmw_ticker.balance_sheet
income = bmw_ticker.income_stmt

returns = pd.DataFrame({
    "ROE (%)":(income.loc["Net Income Common Stockholders"]/
               balance.loc["Stockholders Equity"] * 100).round(1),
    "ROA (%)":(income.loc["Net Income Including Noncontrolling Interests"]/
               balance.loc["Total Assets"] * 100).round(1),
}).dropna(how="all")

#pandas automatically lines up divisions by year
#shareholders profit is shareholders equity for ROE
#this gives us the table of ratios

returns.index = returns.index.year
returns = returns.sort_index()
returns
```

```python
roe_check = (income.loc["Net Income"] / balance.loc["Stockholders Equity"]
             * 100).round(1)
roe_check
#this now matches up with our expected figures, note the trends are still
#the same, however good to know regardless
```

```python
#We now calculate ROIC:
tax_rate = income.loc["Tax Provision"]/income.loc["Pretax Income"]
nopat = income.loc["Operating Income"] * (1 - tax_rate)
invested = balance.loc["Total Debt"] + balance.loc["Stockholders Equity"] - balance.loc["Cash And Cash Equivalents"]

ROIC = (nopat / invested * 100).round(1)
ROIC.index = ROIC.index.year
ROIC.sort_index()
```

ROE is higher than ROIC. ROE is about 7% in 2025 while ROIC is about 4.4%. That gap is the borrowing effect: debt lifts the return to shareholders above the return on the business itself. 2022 profit however looks inflated by the one-off, so the fall from 2022 overstates the decline. 2023 and 2025 give a fairer picture.

#Is 4.4% ROIC good or bad?
It depends on what BMW pays for its capital. A rough rule is that if ROIC sits below the cost of capital, growth destroys value and doesn't create it. For a large industrial company that cost is often in the region of 7-9%, but that's a rough assumption and not a figure from BMW's report, not a fact. The honest conclusion from the data is that group returns have fallen towards or below what investors would normally require, with the caveat that the finance arm distorst all the group ratios.
