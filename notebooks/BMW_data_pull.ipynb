```python
bmw_balance = bmw_ticker.balance_sheet
#now we have the file  of BMW data we simply call upon the balance sheet

bmw_balance
#simply shows us the balance sheet
```

```python
bmw_cashflow = bmw_ticker.cash_flow
#using the exact same logic we can get the cash flow

bmw_cashflow
#and then call up the table for it
```

```python
bmw_revenue = bmw_income.loc["Total Revenue"]
#this isolates the key row we're looking for of Total Revenue from our income statement

bmw_net_income = bmw_income.loc["Net Income"]
#exactly the same now for Net Income

bmw_operating_cash_flow = bmw_cashflow.loc["Operating Cash Flow"]
#Operating cash flow is in cashflow not income statement, but same technique.

print(bmw_revenue)
print(bmw_net_income)
print(bmw_operating_cash_flow)
#we use these commands to then give us the results
```

#What are BMW’s three biggest assets? What do they tell you about its business?

Three biggest assets. All figures are €bn for 2025:

Net PPE, 92.9 (35% of assets). This is the factories, equipment and, probably, cars BMW leases to customers through its finance arm (unconfirmed).

Accounts receivable, 39.4 (15%). At a carmaker with a finance business, this likely includes customer and dealer finance receivables as well as ordinary trade bills.

Inventory, 21.3 (8%). This is cars and parts waiting to be sold, and it fell 13% in 2025, in line with lower revenue.

Together they show a business that’s capital-heavy, with a large finance operation sitting inside it.

#How much of the balance sheet is funded by debt, and how much by equity? Compare Total Debt and Stockholders Equity for the latest year.

Debt versus equity. Total debt is 88.4 and stockholders’ equity is 95.7, so debt to equity is about 0.92. Over the four years it was 0.82 (2022), 0.80 (2023), 0.93 (2024) and 0.92 (2025). Debt jumped by about €14bn in 2024 while equity rose only about €3bn, and I don’t know why from this table. That’s a good question to take to the report’s financial position section (page 69 onwards).

#Is equity growing over the years shown? Is that a sign of a company retaining profits?

Yes, every year: 87.1, 89.6, 92.3, 95.7. That’s consistent with BMW keeping part of its profits after paying dividends and buying back shares. Currency moves and pension changes also affect equity, so don’t read it as profit retention alone.

A reconciliation worth noticing. Liabilities (168.1) plus stockholders’ equity (95.7) is 263.8, not the 266.0 total assets. The 2.2 gap is minority interest, the share of subsidiaries owned by outsiders. That’s why the identity check earlier used “Total Equity Gross Minority Interest” and came out at exactly zero.

BMW’s consolidated debt is mostly funding its finance arm’s loan book, so debt to equity of 0.92 looks worse than the car business alone. BMW reports Automotive and Financial Services separately on page 273, and the picture looks different there. Will need to keep this in mind when choosing ratios.
